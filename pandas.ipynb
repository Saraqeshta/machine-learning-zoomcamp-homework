```python
import numpy as np
import pandas as pd
# مكتبة باندا التعامل مع الجدوال وسلاسل 
```

```python
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
df= pd.DataFrame(data,columns=columns)# نعمل جدول كل ليستا عبارة عن خصائص والاعمدة اسما الخصائص 
df.index=['a','b','c','d','e']#  من الاساس الجدول يكون مرقم من صفر الي العدد ممكن تغير الاندكس 
df.loc['b']# الوصول للعنصر عن طريق اسم الاندكس 
df=df.reset_index(drop=True)# بحال تم تسمية الاندكس وبدنا نرجعه لتريب الارقام من اول وجديد وبتم حفضظ الاندكس القديم في عمود واذا بدك بتحذفه 
df.iloc[1]# الوصول للعنصر عن طريق رقم الاندكس 
df[['Make','Model','MSRP']]# طباعة محتويات الاعمدة المذكورة 
df.loc[0:2,['Make','Model']]# طباعة الصفوف من صف الاول الي تاني والاعمدة المذكورة اسمها 
df['id']=[1,2,3,4,5]# اضافة عمود جديد 
df.head(3)# طباعة اول تلات سطور من جدول 
del df['id']# حذف العمود 
df['Engine HP']*2# تنفيذ عملية رياضية علي العمود 
df['Year']>=2015# فليترات العمود يعني جيب القيم الي اكبر او تساوي كذا 
df[df['Make']=='Nissan']# من العمود كذا جيب الي قيمتها نيسان 
df[(df['Make']=='Nissan')&(df['Year']>=2015)]# اضافة شرطين من العمود كد جيب كدا و كدا 
df['Vehicle_Style'].str.lower() # هان التعامل مع النصوص العمود هادا نحولها النص من تم نخلي كل الحروف سمول 
df['Vehicle_Style'].str.replace(' ','_')# نفس الاشي لكن نحول من وجود مسافة الي ـ
df['Vehicle_Style']=df['Vehicle_Style'].str.replace(' ','_').str.lower() # وهان خلينا العمليتين الي فوق في بعض وكمان حفظهم في متغير الاساسي عشان يتم تعديل عليهم بالجدول 
df['MSRP'].mean()# بجيب المتوسط 
df['MSRP'].max()# اكبر قيمة 
df['MSRP'].min()# اصغر قيمة 
df['MSRP'].std()# الانحراف المعياري 
df['MSRP'].describe()# المعلومات الاحصائية 
df['MSRP'].describe().round(2)# المعلومات الاحصاية تقريب الاتنين 
df.isnull()# نشوف عدد القيم النان المفقودة 
df.isnull().sum()# نجمع عدد القيم المفقودة 
df['Engine HP'].fillna(0)# هان بنحط العمود الي فيه قيم مفقودة وبنعوضها بصفر 
df['Engine HP'].fillna(df['Engine HP'].mean())# نفس الاشي لكن بنحط متوسط الاجابات 
df.dropna()# او ممكن نحذف العمود او صف الي في قيمة مفقودة 
df.groupby('Transmission Type')['MSRP'].mean()# هان تقسيم العملية الي مجموعة عمليات بنا علي العمود المعني يعني هان بقوله العمود انواع المحركات كم متوسط سعره بنا علي نوع المحرك 
df['Make'].nunique()# عدد القيم الفريدة 
df['Make'].unique()#قائمة القيم المفقودة 
df['Make'].value_counts()# عدد تكرار كل قيمة 
df.nunique()# قائمة كل القيمة الفريدة 
df['MSRP'].values# تحويل محتوي العمود الي مصفوفة 
df.to_dict(orient='records')# حفظ الجدول في قاموس كل واحد لوحده 
```
